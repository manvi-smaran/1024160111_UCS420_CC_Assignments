```python
import pandas as pd
```

Q1

```python
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = [
    {
        "question": "how can i get a refund for an incorrect payment",
        "answer": "Contact the billing department with your payment details to request a refund.",
        "keywords": "refund payment billing",
        "category": "billing"
    },
    {
        "question": "how do i update my registered mobile number",
        "answer": "Go to Account Settings and update your registered mobile number.",
        "keywords": "mobile number update account",
        "category": "account"
    }
]

faq_entries = fixed_entries + personalized_entries
df = pd.DataFrame(faq_entries)
print(df)
```

Q2

```python
def score_query(query, df):
    results = []
    query_words = set(query.lower().split())
    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())
        matches = query_words & text_words
        if matches:
            confidence = len(matches) / len(query_words)
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })
    return pd.DataFrame(results).sort_values("confidence", ascending=False)

query = "how can I pay my fee"
print(score_query(query, df))
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

result = same_category("billing", df)
print(result)
```

```python
def score_query(query, df):
    results = []
    query_words = set(query.lower().split())
    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        text_words = set(text.split())
        matches = query_words & text_words
        if matches:
            confidence = len(matches) / len(query_words)
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })
    if not results:
        print("No matching entries.")
        return
    highest = max(x["confidence"] for x in results)
    best_matches = [x for x in results if x["confidence"] == highest]
    print("Best match(es):")
    for x in best_matches:
        print(x)

print("Query 1:")
score_query("fee", df)
print("\nQuery 2:")
score_query("password", df)
```
