```python
VemulaManviSmaran = np.linspace(10, 100, 25)

print(f"Dimensions: {VemulaManviSmaran.ndim}")
print(f"Shape: {VemulaManviSmaran.shape}")
print(f"Total elements: {VemulaManviSmaran.size}")
print(f"Data type: {VemulaManviSmaran.dtype}")
print(f"Bytes consumed: {VemulaManviSmaran.nbytes}")
print("Array content:")
print(VemulaManviSmaran)
```

```python
transposed_with_reshape = VemulaManviSmaran.reshape(-1, 1)
print("Transpose using reshape(-1, 1):")
print(transposed_with_reshape)
print(f"Shape after reshape: {transposed_with_reshape.shape}")
```

```python
print("Transpose of reshaped array using .T attribute:")
print(transposed_with_reshape.T)
print(f"Shape of transposed array using .T: {transposed_with_reshape.T.shape}")

print("\n.T attribute on original 1-D array:")
print(VemulaManviSmaran.T)
print(f"Shape of original array using .T: {VemulaManviSmaran.T.shape}")
```

```python
ucs420_VemulaManviSmaran = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_VemulaManviSmaran )
print(f"Shape: {ucs420_VemulaManviSmaran .shape}")

print(f"\nMean: {np.mean(ucs420_VemulaManviSmaran )}")
print(f"Median: {np.median(ucs420_VemulaManviSmaran )}")
print(f"Max: {np.max(ucs420_VemulaManviSmaran )}")
print(f"Min: {np.min(ucs420_VemulaManviSmaran )}")
print(f"Unique elements: {np.unique(ucs420_VemulaManviSmaran )}")
```

```python
reshaped_ucs420_VemulaManviSmaran  = ucs420_VemulaManviSmaran .reshape(4, 3)

print("Reshaped Array (4 rows, 3 columns):")
print(reshaped_ucs420_VemulaManviSmaran )
print(f"Shape: {reshaped_ucs420_VemulaManviSmaran .shape}")
```

```python
resized_ucs420_VemulaManviSmaran  = ucs420_VemulaManviSmaran .flatten()[:6].reshape(2, 3)

print("Resized Array (2 rows, 3 columns):")
print(resized_ucs420_VemulaManviSmaran )
print(f"Shape: {resized_ucs420_VemulaManviSmaran .shape}")
```
