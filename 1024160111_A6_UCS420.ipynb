```python
# Q1(a-e): Sensor readings, vectorization and Boolean indexing

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

c = temperature + 2
f = (9/5) * c + 32
x = c[c > 32]
n = np.sum(c > 32)

print("Corrected Celsius:", c)
print("Fahrenheit:", f)
print("Readings > 32°C:", x)
print("Count > 32°C:", n)

print("\nVectorization applies operations to the complete array at once,")
print("which is generally faster and simpler than explicitly using a loop.")
print("Boolean indexing directly selects elements satisfying a condition.")
```

```python
# Q2(a-f): Analyze daily step data using NumPy

steps = np.array([[5000, 6200, 7100],
                  [8000, 7500, 9000],
                  [4500, 5100, 4800],
                  [9000, 8500, 9500]])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
print("Total for each day:", np.sum(steps, axis=0))
print("Total for each user:", np.sum(steps, axis=1))

p = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position (user, day):", p)
```

```python
# Q3(a-j): NumPy slicing, copying, reshaping, flattening and array properties

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copy_a = original[1:5].copy()
copy_a[0] = 500

print("Original after modifying copy:", original)
print("Copied array:", copy_a)

a = np.arange(1, 13).reshape(3, 4)

print("\nMatrix:\n", a)
print("First row:", a[0])
print("Last row:", a[-1])
print("Second column:", a[:, 1])
print("Rows 1-2, columns 2-3:\n", a[1:3, 2:4])

flat = a.flatten()
rav = a.ravel()

rav[0] = 100
print("\nAfter modifying ravel:")
print("Original:\n", a)
print("Ravel:", rav)

flat[1] = 200
print("\nAfter modifying flatten:")
print("Original:\n", a)
print("Flatten:", flat)

print("\nShape:", a.shape)
print("Dimensions:", a.ndim)
print("Size:", a.size)
print("Data type:", a.dtype)
```

```python
# Q4(a-g): Matrix operations, OLS coefficients and prediction

X = np.array([[6, 70, 3],
              [5, 50, 6],
              [8, 80, 2],
              [4, 30, 8]])

y = np.array([40, 65, 30, 85])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

xt = X.T
print("\nX.T:\n", xt)

print("\nX.T @ X:\n", xt @ X)

b = np.linalg.inv(xt @ X) @ xt @ y
print("\nCoefficients:", b)

new_user = np.array([5, 40, 7])
pred = new_user @ b

print("\nPredicted assistance score:", pred)

print("\nThe coefficients represent the estimated effect of")
print("sleep hours, activity level and stress level on the assistance score.")
```
