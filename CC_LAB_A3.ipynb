```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("employees.csv", index=False)

print(df)
```

```python
df=pd.read_csv("employees.csv")
print(df.shape)
print(df.info())
print(df.describe())
print(df.head(5))
print(df.tail(3))
print("Average Salary:", df["Salary"].mean())
print("Total Bonus:",df["Bonus"].sum())
print("Youngest Employee:",df["Age"].min())
print("Highest Performance Rating:",df["Rating"].max())
df = df.sort_values(by="Salary", ascending=False)
print("\nDataFrame sorted by Salary:")
print(df)
```

```python
def performance_cat(rating):
  if rating>=4.5:
    return "Excelent"
  elif rating>=4.0:
    return "Good"
  else:
    return "Average"

df["Performance"]=df["Rating"].apply(performance_cat)
print("\nPerformance Category:")
print(df[["Name", "Rating", "Performance"]])
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.rename(columns={"Employee_ID":"ID"},inplace=True)
print("\nAfter renaming Employee_ID to ID:")
print(df)
```

```python
print("\nEmployees with more than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])

print("\nEmployees belonging to IT department:")
print(df[df["Department"] == "IT"])

df["Tax"] = df["Salary"] * 0.10

print("\nDataFrame with Tax:")
print(df)

df.to_csv("modified_employees.csv", index=False)
print("\nModified dataset saved as modified_employees.csv")
```
